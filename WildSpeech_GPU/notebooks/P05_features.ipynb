# P05 — Student-level feature engineering
Recordings are never treated as independent observations: every row is a student. Per task and globally: item-normalised acoustic statistics (mean/std/median/p10/p90), correct-vs-incorrect contrasts, complexity (difficulty) sensitivity slopes, behaviour (items, unscored fraction, durations), difficulty-adjusted accuracy, fold-fitted Rasch theta, linguistic load, screening and recording covariates.

This stage writes **cross-fitted** features (each row from a builder fitted without that student) for the discovery stages P06–P09. Predictive stages rebuild features inside every CV fold.
